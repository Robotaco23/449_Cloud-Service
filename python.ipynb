## Part 1 - Slides 21-23

Calculate the mean of values: 1, 2, 5

```python
values = [1, 2, 5]
sum(values) / len(values)
```

Calculate the magnitude of vector (1, 2, 5)

Magnitude = $\sqrt{1^2 + 2^2 + 5^2}$

```python
vector = [1, 2, 5]
magnitude = (1**2 + 2**2 + 5**2) ** 0.5
magnitude
```

Normalize the vector

*Divide each component by the magnitude (and save to a new variable)*

```python
normalized_vector = [x / magnitude for x in vector]
normalized_vector
```

Calculate the magnitude of the normalized vector

```python
normalized_magnitude = sum(x**2 for x in normalized_vector) ** 0.5
normalized_magnitude
```


Calculate the mean of 1st 100 numbers: 1, 2, …, 100

```python
numbers = list(range(1, 106))
mean_of_numbers = sum(numbers) / len(numbers)
print(f"The mean of the first 100 numbers is: {mean_of_numbers}")
```
